## ES 3

```python
stringa = input("Inserisci una stringa:")
new_stringa = ""
if not stringa:
    print("errore")
else:
    for i in range(len(stringa)):
        if i % 2 == 0:
            new_stringa += stringa[i]
    print(new_stringa)
```
