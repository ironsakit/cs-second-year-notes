Speriamo di sopravvivere

```python
stringa = "ciao"
print(stringa[2:])
print(stringa[:2])
print(stringa[::-1])
```

```python
print(5 / 2)
print(5 // 2)
print(5 % 2)
print(5 ** 2)
```

## Type | '_' | basi diverse
> Permette di visualizzare il tipo della variabile

```python
print(type(2.0))
```

```python
print(1_000_134)   # '_' per separare le cifre
```

```python
print(1e3)        # notazione scientifica
```

```python
print(0xFF)    # esadecimale
print(0o33)    # ottale
print(0b11)    # binario
```

## Tuple | Liste

```python
t = (1,2,3)    # tupla: immutabile
print(t)
# t[0] = 9  --> non posso farlo

l = [1,2,3]   # lista: mutabile
print(l)
l[0] = 9
print(l)
```
